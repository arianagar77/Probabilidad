#**Interpretaciones de la Probabilidad**

A lo largo del desarrollo de la teoría de la probabilidad han surgido distintas interpretaciones sobre qué significa realmente una probabilidad. Aunque todas comparten el mismo marco matemático, difieren en la forma en que interpretan y justifican el valor de una probabilidad.


Las tres interpretaciones más influyentes en estadística y ciencia de datos son:

- interpretación clásica

- interpretación frecuentista

- interpretación bayesiana

#**Clasica**

La **interpretación clásica** es históricamente la primera y surge en el estudio de los juegos de azar durante los siglos XVII y XVIII, en trabajos de matemáticos como Pascal y Fermat.


En este enfoque, la probabilidad de un evento se define como la razón entre el número de resultados favorables y el número total de resultados posibles, suponiendo que todos los resultados son igualmente probables.


Este enfoque es muy útil para introducir los conceptos básicos, aunque su aplicación se limita principalmente a experimentos con espacios muestrales bien definidos y equiprobables.

##**Probabilidad Clásica**

La probabilidad de un evento
A, P(A) es un número real que toma valores en el intervalo
[
0
,
1
]
[0,1]. Este valor representa una medida de qué tan probable es que ocurra el evento cuando se realiza el experimento aleatorio correspondiente.

En términos intuitivos, puede interpretarse como una medida de qué tan frecuentemente esperamos observar ese evento si el experimento se repite muchas veces bajo las mismas condiciones.

#Ejemplo 2

```python
from itertools import product

S = list(product(range(1,7), repeat=2))
print("Espacio Muestral:", S)

A = [x for x in S if x[0] + x[1] == 7]
print("Casos favorables:", A)

print("Probabilidad:", len(A)/len(S))
```

#Ejemplo 3

```python
from itertools import permutations

libros = ["A", "B", "C"]

S = list(permutations(libros, 2))
print("Espeacio Muestral:")
for p in S:
    print(p)

# evento: primer libro es A
evento = [x for x in S if x[0] == "A"]
print("Evento:", evento)

print("Probabilidad:", len(evento)/len(S))
```

#Ejemplo 4

```python
from itertools import product

S = list(product(range(10), repeat=3))
print("Espacio Muestral:", S)

A = [x for x in S if x[0] == x[1] == x[2]]
print("Casos favorables:", A[:10])  # solo muestra algunos

print("Probabilidad:", len(A)/len(S))
```

#Ejemplo 5

```python
from itertools import product

S = list(product(["A","S"], repeat=3))
print("Espacio Muestral:", S)

A = [x for x in S if x.count("A") == 2]
print("Casos favorables:", A)

print("Probabilidad:", len(A)/len(S))
```

#Ejercicio 1

Al lanzar dos monedas al aire, calcula la probabilidad de que ambas caigan en la misma cara (ambas águila o ambas sol)

```python
from itertools import product

S = list(product(["Aguila","Sol"], repeat=2))
print("Espacio Muestral:", S)

A = [x for x in S if x[0] == x[1]]
print("Casos favorables:", A)

print("Probabilidad:", len(A)/len(S))
```

#Ejercicio 2

Al lanzar dos dados, calcula la probabilidad de que la suma de sus caras sea exactamente 5.

```python
from itertools import product

S = list(product([str(i) for i in range(1,7)], repeat=2))
print("Espacio Muestral:", S)

A = [x for x in S if int(x[0]) + int(x[1]) == 5]
print("Casos favorables:", A)

print("Probabilidad:", len(A)/len(S))
```
