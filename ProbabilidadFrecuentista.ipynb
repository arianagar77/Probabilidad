#**Frecuentista**

La interpretación frecuentista entiende la probabilidad como el límite de la frecuencia relativa de un evento cuando el experimento se repite muchas veces bajo las mismas condiciones.

Este enfoque es especialmente importante en estadística y ciencia de datos, ya que muchas probabilidades se estiman a partir de datos observados.

#**Probabilidad frecuentista**

La probabilidad de un evento puede interpretarse como el valor al que tiende la frecuencia relativa de ese evento cuando el experimento se repite un número muy grande de veces.

Al repetir el experimento un número suficientemente grande de veces es posible obtener aproximaciones cada vez más cercanas al valor real de la probabilidad. Aunque esta definición no es completamente formal desde el punto de vista matemático, resulta muy útil para interpretar la probabilidad a partir de datos observados.

##**Ejemplo 1**

¿Cuál es la probabilidad de obtener “cara” al lanzar una moneda?

```python
import random
import matplotlib.pyplot as plt

# Funcion que simula el lanzamiento de una moneda
def lanzar_moneda():
    # Elegimos entre "águila" y "sol"
    return random.choice(["águila", "sol"])

# Numero total de lanzamientos
n = 10000
# Contador de aguilas
conteo_aguilas = 0
# Lista para guardar la frecuencia relativa
frecuencias = []

# Repetimos el experimento n veces
for i in range(1, n + 1):
    resultado = lanzar_moneda()
    # Verificamos si salio águila
    if resultado == "águila":
        conteo_aguilas += 1
    # Frecuencia relativa
    frecuencia = conteo_aguilas / i
    frecuencias.append(frecuencia)

print("Número de lanzamientos:", n)
print("Probabilidad empírica de obtener águila:", frecuencia)
```

```python
# Creamos una figura para la gráfica
plt.figure(figsize=(10,5))
# Graficamos como cambia la frecuencia relativa conforme aumentan los lanzamientos
plt.plot(frecuencias, label="Frecuencia relativa de águila")
# Dibujamos una línea horizontal en 0.5 (valor teorico esperado)
plt.axhline(0.5, linestyle="--", label="Probabilidad teórica = 1/2")
# Etiquetas de los ejes
plt.xlabel("Número de lanzamientos")
plt.ylabel("Frecuencia relativa")
# Título de la grafica
plt.title("Convergencia de la probabilidad empírica al valor teórico")
plt.legend()
# Mostramos la grafica
plt.show()
```

##**Ejemplo 2**
Corresponde a obtener un número par al lanzar un dado.

```python
import random
import matplotlib.pyplot as plt

# Definimos una funcion que simula el lanzamiento de un dado
def lanzar_dado():
    # randint(1,6) genera un numero entero aleatorio entre 1 y 6
    return random.randint(1,6)

# Numero total de lanzamientos del experimento
n = 10000
# Variable para contar cuantas veces aparece un numero par
conteo_pares = 0
# Lista donde guardaremos la frecuencia relativa despues de cada lanzamiento
frecuencias = []

# Repetimos el experimento n veces
for i in range(1, n+1):
    # Simulamos el lanzamiento del dado
    resultado = lanzar_dado()
    # Verificamos si el resultado es par
    if resultado % 2 == 0:
        conteo_pares += 1
    # Calculamos la frecuencia relativa del evento
    # numero de pares observados / numero total de lanzamientos
    frecuencia = conteo_pares / i
    # Guardamos este valor en la lista
    frecuencias.append(frecuencia)

print("Número de lanzamientos:", n)
print("Probabilidad empírica de obtener un número par:", frecuencia)
```

```python
# Creamos una figura para la gráfica
plt.figure(figsize=(10,5))
# Graficamos cómo cambia la frecuencia relativa conforme aumentan los lanzamientos
plt.plot(frecuencias, label="Frecuencia relativa de números pares")
# Dibujamos una línea horizontal en 0.5
plt.axhline(0.5, linestyle="--", label="Probabilidad teórica = 1/2")
# Etiquetas de los ejes
plt.xlabel("Número de lanzamientos")
plt.ylabel("Frecuencia relativa")
# Título de la gráfica
plt.title("Convergencia de la probabilidad empírica al valor teórico")
plt.legend()
# Mostramos la gráfica
plt.show()
```

##**Ejemplo 3**

Una urna tiene 3 rojas y 2 azules. ¿Probabilidad de sacar una roja?

```python
import random
import matplotlib.pyplot as plt

# Definimos una funcion que simula el lanzamiento de un dado
def sacar_bola():
    urna = ["roja", "roja", "roja", "azul", "azul"]
    # Elegimos entre "roja" y "azul"
    return random.choice(urna)

# Numero total de lanzamientos del experimento
n = 10000
# Variable para contar cuantas veces obtenemos una bola roja
conteo_rojas = 0
# Lista donde guardaremos la frecuencia relativa despues de cada lanzamiento
frecuencias = []

# Repetimos el experimento n veces
for i in range(1, n+1):
    # Simulamos el lanzamiento del dado
    resultado = sacar_bola()
    # Verificamos si el resultado es par
    if resultado == "roja":
        conteo_rojas += 1
    # Calculamos la frecuencia relativa del evento
    # numero de pares observados / numero total de lanzamientos
    frecuencia = conteo_pares / i
    # Guardamos este valor en la lista
    frecuencias.append(frecuencia)

print("Número de lanzamientos:", n)
print("Probabilidad empírica de obtener una bola roja:", frecuencia)
```

```python
# Creamos una figura para la gráfica
plt.figure(figsize=(10,5))
# Graficamos cómo cambia la frecuencia relativa conforme aumentan los lanzamientos
plt.plot(frecuencias, label="Frecuencia relativa de obtener una bola roja")
# Dibujamos una línea horizontal en 0.5
plt.axhline(3/5, linestyle="--", label="Probabilidad teórica = 3/5")
# Etiquetas de los ejes
plt.xlabel("Número de lanzamientos")
plt.ylabel("Frecuencia relativa")
# Título de la gráfica
plt.title("Convergencia de la probabilidad empírica al valor teórico")
plt.legend()
# Mostramos la gráfica
plt.show()
```

##**Ejemplo 4**

Un productor quiere estimar:

¿Cuál es la probabilidad de tener una buena cosecha (> 8 toneladas por hectárea)?

En este escenario, no tenemos fórmula teórica, pero sí conocimiento empírico:


- Lluvia anual (mm)

- Fertilizante usado

- Temperatura promedio


Para esto debemos desarrollar un modelo (simulación) que determine el rendimiento de la cosecha.

Construimos un modelo simple (no perfecto, pero si realista)

donde:

- Lluvia es una variable aleatoria

- Fertilizante es una variable controlada

- Ruido es la incertidumbre (clima, plagas, etc.)

```python
import numpy as np

n = 10000  # numero de simulaciones

# Variables aleatorias
lluvia = np.random.normal(800, 150, n)       # mm/ano
fertilizante = np.random.normal(5, 1, n)     # unidades
ruido = np.random.normal(0, 1, n)            # incertidumbre

# Modelo de rendimiento (toneladas/hectárea)
rendimiento_cosecha = 0.005 * lluvia + 0.8 * fertilizante + ruido

# Evento: buena cosecha (> 8 toneladas)
buena_cosecha = rendimiento_cosecha > 8

# Probabilidad frecuentista
prob = np.sum(buena_cosecha) / n

print("Probabilidad de buena cosecha ≈", prob)
```

```python
# Frecuencia acumulada
frecuencia = np.cumsum(buena_cosecha) / np.arange(1, n + 1)
# Gráfica
plt.plot(frecuencia, label="Frecuencia acumulada")
plt.title("Convergencia de la probabilidad de buena cosecha")
plt.xlabel("Número de simulaciones")
plt.ylabel("Probabilidad")
plt.legend()
plt.show()
```
